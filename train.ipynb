```python
print("Dataset location:", dataset.location)
```

```python
model = YOLO("yolo11m.pt")
```

```python
results = model.train(
    data=f"{dataset.location}/data.yaml",

    epochs=150,
    imgsz=640,
    batch=8,

    device=0,

    project="clamp_training",
    name="yolov11m",

    patience=30,

    optimizer="AdamW",
    lr0=0.001,

    pretrained=True
)
```
